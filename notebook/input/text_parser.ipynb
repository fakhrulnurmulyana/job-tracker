```python
def extract_brackets_one(text):
    results = []
    depth = 0
    start = -1

    for i, char in enumerate(text):
        if char == '{':
            if depth == 0:
                start = i  # mark position of first {
            depth += 1
        elif char == '}':
            depth -= 1
            if depth == 0 and start != -1:
                results.append(text[start:i+1])  # grab from { to its matching }
                start = -1

    return results

def extract_brackets(contents):
    return [
        extract_brackets_one(content)
        for content in contents
    ]
```

```python
def match_keys_one(contents, keys):
    result = dict()
    for i, content in enumerate(contents):
        for j, (key, value) in enumerate(content.items()):
            if key in keys:
                result[key]=value
    
    return result

def match_keys(contents, keys):
    result = {}
    for content in contents:
        result |= match_keys_one(content, keys)
    return result
```
