```python
import json
import sys

from pathlib import Path
from typing import List, Sequence

current = Path().resolve().parent.parent.parent
source_code = current / "src"

sys.path.append(str(source_code))

from job_tracker.infrastructure import PathResolver
from job_tracker.orchestration import DistributionChart
```

```python
path_resolver = PathResolver(base_path=current)
distribution_chart = DistributionChart()

data_path = path_resolver.finalized_directory()
```

```python
json_data = []

for file in data_path.rglob("*.json"):
    with open(file, "r", encoding="utf-8") as f:
        json_data.append(json.load(f))

all_length = len(json_data)

print(all_length)
```

```python
all_keys = json_data[1].keys()
```

```python
all_data = dict()

for key in all_keys:
    context_doc = dict()
    for context in json_data[1][key].keys():
        context_doc[context] = ""
    all_data[key] = context_doc
```

```python
def normalizer_salary(salary: int | float):
    result=list()
    for money in salary:
        if money is None:
            result.append(None)
        else:
            operation = money / 1000000
            result.append(operation)

    return result
```

```python
def normalization_salary_viz(data:Sequence[float])->List[str]:    
    result = [
        f"{x:.2f}M"
        for x in sorted(
            x for x in data if x is not None
        )
    ]

    return result
```

# Job

```python
job_keys = json_data[1]["job"].keys()
job_keys
```

```python
job_doc={
    f"job_{key}": list() 
    for key in job_keys
    }

for data in json_data:
    for key in job_keys:
        context = f"job_{key}"
        job_doc[context].append(data["job"][key])
```

## salary

```python
job_doc["job_salary"]
```

```python
temp_job_salary = {key: list() for key in job_doc["job_salary"][0].keys()}

for detail_salary in job_doc["job_salary"]:
    for key, value in (detail_salary).items():
        temp_job_salary[key].append(value)

job_doc["job_salary"] = temp_job_salary
```

```python
job_doc["job_salary"]["max"]
```

```python
for threshold in ["min", "max"]:
    job_doc["job_salary"][threshold] = normalizer_salary(job_doc["job_salary"][threshold])
```

```python
distribution_chart.distribution_chart(
    data=job_doc["job_salary"]["displayed"],
    chart_type="pie",
    title_text="Job Displayed Category",
)
```

```python
viz_min_job_salary = normalization_salary_viz(data=job_doc["job_salary"]["min"])
viz_max_job_salary = normalization_salary_viz(data=job_doc["job_salary"]["max"])

distribution_chart.distribution_chart(
    names= [
        'Job Minimal Salary (Millions)', 
        'Job Maximal Salary (Millions)',
    ],
    data= [
        viz_min_job_salary,
        viz_max_job_salary,
    ],
    chart_type="batch_box_plot",
    title_text="Box Plot Styling Outliers",
)
```
